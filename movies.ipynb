```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv("tmdb_5000_credits.csv")
```

```python
movies = movies.merge(credits,on = 'title')
```

```python
# genres
# id
# keywords
# title
# overview
# cast
# crew
movies = movies[["movie_id","genres","id","keywords","title","overview","cast","crew"]]
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace = True)
```

```python
movies.duplicated().sum()
```

```python
import ast
def rearange(obj):
    list = []
    for i in ast.literal_eval(obj):
        list.append(i['name'])
    return list
```

```python
movies["genres"] = movies["genres"].apply(rearange)
movies["keywords"] = movies["keywords"].apply(rearange)
```

```python
def convert(obj):
    list = []
    a = 3
    for i in ast.literal_eval(obj):
        if a <= 0:
            break
        list.append(i['character'])
        a = a-1
    return list
```

```python
movies['cast'] = movies['cast'].apply(convert)
```

```python
def direct(obj):
    list = []
    for i in ast.literal_eval(obj):
        if i['job'] == "Director":
            list.append(i["name"])
    return list
```

```python
movies["crew"] = movies["crew"].apply(direct)
```

```python
movies["overview"][0]
```

```python
movies["overview"] = movies["overview"].apply(lambda x:x.split())
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview']+movies['genres']+ movies['keywords'] + movies['cast'] + movies['crew'] 
```

```python
df = movies[['movie_id','title', 'tags']]
```

```python
df["tags"] = df["tags"].apply(lambda x:" ".join(x))
```

```python
df["tags"][0]
```

```python
df['tags'] = df['tags'].apply(lambda x: x.lower())
```

```python
df.head()
```

```python
 import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def super(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
df["tags"] = df["tags"].apply(super)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv  = CountVectorizer(max_features = 5000,stop_words = 'english')
```

```python
vectors = cv.fit_transform(df["tags"]).toarray()
```

```python
vectors[0]
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
    movie_index = df[df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)),reverse = True,key = lambda x:x[1])[1:6]
    for i in movies_list:
        print(df.iloc[i[0]].title)
```
